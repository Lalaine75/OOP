```python
class BankAccount:
    def __init__(self, account_number, balance):
        self.__account_number = account_number
        self.__balance = balance
        self.frozen = False
        pass

    # ----- read-only getters via @property (NO @property.setter!) -----
    @property
    def account_number(self):
        return self.__account_number

    @property
    def balance(self):
        return self.__balance

    # ----- traditional setter METHODS (called WITH parentheses) -----
    def set_account_number(self, account_number):
        self.__account_number = account_number

    def set_balance(self, balance):
        # check if bank account is frozen
        if self.frozen:
            print("Error: Cannot set balance while bank account is frozen.")
            return

        print(f"Update balance to {balance}")
        # prevent quantum physicists from making another paradox
        if balance < 0:
            print("Error: balance should not be negative")
            return

        self.__balance = balance
        # if sussy, then freeze
        # sussy threshold: 50000 money or greater
        if balance >= 50000:
            print("Warning: suspicious transaction. Bank account has been frozen.")
            self.frozen = True


# ------------------------------------------------------------
#  Step 6 — Test your class.
#  Uncomment and complete the lines below once your class works.
# ------------------------------------------------------------
acct = BankAccount("ACC1", 500)

print(acct.account_number)   # Expected: ACC1   (read — no parentheses)
print(acct.balance)          # Expected: 500    (read — no parentheses)

acct.set_balance(800)        #                 (change via the METHOD)
print(acct.balance)          # Expected: 800

acct.set_balance(-50)        # Expected: a warning is printed
print(acct.balance)          # Expected: 800    (negative was IGNORED!)

# Bonus stuff
acct.set_balance(50000)      # Excepted: suspicious transaction. Bank account has been frozen.
print(acct.balance)          # Expected: 50000    (deposits may still work)

acct.set_balance(800)        # Expected: Cannot set balance while bank account is frozen.
print(acct.balance)          # Expected: 50000
```

```python
#Level 1: Base parent class
class Animal:
  def __init__(self, kingdom = "Animalia"):
    self.kingdom = kingdom

  def eat(self):
    print("Consuming food for energy.")

#Level 2: Child of Animal and Parent of Dog
class Mammal(Animal):
  def __init__(self, kingdom, warm_blooded: True):
    super().__init__(kingdom)  #passes kingdom up to Animal constructor
    self.warm_blooded = warm_blooded

  def nurse_young(self):
    print("Feeding milk to offspring.")

#Sub class of Mammal
class Dog(Mammal):
  def __init__(self, kingdom, warm_blooded, breed, name):
    super().__init__(kingdom, warm_blooded) #Ask Mammal to handle kingdom and warm_blooded initialization
    self.breed = breed
    self.name = name

  def bark(self):
    print(f"{self.name} the {self.breed} says Woof!")

#Testing the Multi-level Biology Chain
my_dog = Dog("Animalia", True, "Shih Tzu", "Bruno")

#Accessing the attributes inherited from 3 levels
print("Kingdom: ", my_dog.kingdom)
#print(f"Kingdom: {my_dog.kingdom})

print("Warm-blooded: ", my_dog.warm_blooded)
print("Breed: ", my_dog.breed)

#Executing methods inherited from all 3 levels
my_dog.eat()
my_dog.nurse_young()
my_dog.bark()



```

##Hospital Billing System

```python
# Parent Class
class Patient:
    def __init__(self, name, bill_amount):
        self.__name = name
        self.__bill_amount = bill_amount

    # Getter and Setter for name
    def get_name(self):
        return self.__name

    def set_name(self, name):
        self.__name = name

    # Getter and Setter for bill amount
    def get_bill_amount(self):
        return self.__bill_amount

    def set_bill_amount(self, amount):
        if amount >= 0:
            self.__bill_amount = amount
        else:
            print("Bill amount cannot be negative.")

    # Method to calculate full bill
    def calculate_bill(self):
        return self.__bill_amount

    # New method that can be overridden
    def display_patient_type(self):
        return "Regular Patient"


# Child Class: Senior Citizen (20% discount)
class SeniorPatient(Patient):
    def calculate_bill(self):
        original_bill = super().calculate_bill()
        discount = 0.20
        return original_bill * (1 - discount)

    def display_patient_type(self):
        return "Senior Patient"


# Child Class: PhilHealth Member (15% discount)
class PhilHealthPatient(Patient):
    def calculate_bill(self):
        original_bill = super().calculate_bill()
        discount = 0.15
        return original_bill * (1 - discount)

    def display_patient_type(self):
        return "PhilHealth Patient"


# Instantiate objects
regular_patient = Patient("Juan Dela Cruz", 15000)
senior_patient = SeniorPatient("Maria Santos", 15000)
philhealth_patient = PhilHealthPatient("Pedro Reyes", 15000)

# Store objects in a list
patients = [regular_patient, senior_patient, philhealth_patient]

# Display results
print("Hospital Billing System")
print("-----------------------")

for patient in patients:
    print("Patient Name: ", patient.get_name())
    print("Patient Type: ", patient.display_patient_type())
    print(f"Final Bill: ₱{ patient.calculate_bill():,.2f}\n")
```
