```python
#----SMT Fair Badge & Station Assignment-----

#input data
name = input("Enter student name: ")
grade = int(input("Enter grade level (9-12): "))
needs_electricity = input("Does your project needs electricity (yes/no): ")

#function definitions
def find_location(needs_electricity):
  if (needs_electricity == "yes"):
      assigned_location = "MMC - Power Station"
  else:
    assigned_location = "Gym - General Display"
  return assigned_location

def find_time_slot(grade):
  if grade == 9:
    time_slot = "9:00 AM - 10:30 AM"
  elif grade == 10:
    time_slot = "10:45 AM - 12:15 PM"
  elif grade == 11 or grade == 12:
    time_slot = "1:00 PM - 2:30 PM"
  else:
    time_slot = "Invalid grade level"
  return time_slot


#function call
assigned_location = find_location(needs_electricity)
time_slot = find_time_slot(grade)


#output
print("--------------------------------")
print("OFFIICIAL SMT FAIR BADGE")
print(f"Student: {name}")
print(f"Assigned location: {assigned_location}")
print(f"Time slot: {time_slot}")
```

```python
temperatures = [28, 30, 19, 29, 21, 17]
temperatures.append(32)
highest_temp = max(temperatures)

print(temperatures)
print(f"the highest temp is {highest_temp}")
average_temp = sum(temperatures) / len(temperatures)
print(f"the average temperature: {round(average_temp, 2)} °C")
print(temperatures)
temperatures[2] = 27
print(temperatures)

for reading in temperatures:
  print(reading)
  if reading < 20:
      temperatures.remove(reading)
      print(f"item removed: {reading}")
  else:
      continue

print(temperatures)
```
